```python
%sql
-- =====================================================
-- Notebook: 00_create_catalog_schemas
-- Objectif : Initialiser le Catalog Unity Catalog et les
--            schémas Medallion (Bronze/Silver/Gold)
-- Auteur   : Ahmed Alaeddine Baatour
-- Projet   : Banking Lakehouse - Databricks
-- =====================================================

USE CATALOG banking_lakehouse;

CREATE SCHEMA IF NOT EXISTS bronze
  COMMENT 'Zone Bronze - Données brutes ingérées telles quelles';

CREATE SCHEMA IF NOT EXISTS silver
  COMMENT 'Zone Silver - Données nettoyées, typées, dédupliquées';

CREATE SCHEMA IF NOT EXISTS gold
  COMMENT 'Zone Gold - Données agrégées, modèle en étoile pour BI';

CREATE VOLUME IF NOT EXISTS bronze.landing_files
  COMMENT 'Landing zone pour les fichiers CSV/JSON bruts avant ingestion';

-- Vérification
SHOW SCHEMAS IN banking_lakehouse;
```
